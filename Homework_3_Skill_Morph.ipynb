```python
#Write a program to find the largest and smallest numbers in a tuple

numbers = (10, 20, 40, 60, 21, 50, 124, 856, 13)
Largest_Numbers = max(numbers)
Smallest_Numbers = min(numbers)
print("The highest number is: ", Largest_Numbers)
print("The smallest number is: ", Smallest_Numbers)
```

```python
#Write a function that takes string and returns the reverse of that string. For example "hello" should become "olleh"

def string(text):
  return text[::-1]
print(string("hello"))



```

```python
#Write a function that takes string and returns the reverse of that string. For example "hello" should become "olleh"
def reverse_string(text):
  reverse_text = ""
  for i in text:
    reverse_text = i + reverse_text
  return reverse_text
print(reverse_string("hello"))


```

```python
#Given two sets set1 = {1, 2, 3, 4} and set2 = {3, 4, 5, 6}, write a function that returns a set containing the elements that are in set1 but not in set2.

set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
def difference_set(set1, set2):
  return set1 - set2
result = difference_set(set1, set2)
print(result)
```
